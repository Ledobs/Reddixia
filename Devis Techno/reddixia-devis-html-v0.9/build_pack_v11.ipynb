# Build Reddixia Pack v11

Ce notebook reconstruit le pack HTML v11 à partir des artefacts locaux (ToolboxAgents.xlsx, DiagramCode.txt).

> Note : l'accès internet est désactivé dans l'environnement notebook de ChatGPT. Pour intégrer les contenus GitHub (dat_/index_/architecture.md), cloner le repo localement puis pointer les chemins vers les fichiers `.md`.

```
from pathlib import Path
import pandas as pd, re, json, hashlib, datetime as dt, zipfile, shutil

base = Path('.')
source_dir = Path('.')  # adapter si nécessaire
out_dir = Path('reddixia_pack_v11')

agents_xlsx = source_dir/'ToolboxAgents.xlsx'
diagram_code = source_dir/'DiagramCode.txt'

print('Agents:', agents_xlsx.exists(), 'DiagramCode:', diagram_code.exists())
```

```
# Charger la liste des agents

df_agents = pd.read_excel(agents_xlsx, sheet_name='Listing')
df_agents.columns = [c.strip() for c in df_agents.columns]
for c in ['Domaine','NomAgent','NomCode','CheminGit']:
    if c in df_agents.columns:
        df_agents[c] = df_agents[c].astype(str).str.strip()

print('Agents:', df_agents.shape)
print(df_agents.head(3))
```

```
# Extraire les tables tpg_* depuis DiagramCode.txt

code = diagram_code.read_text(encoding='utf-8', errors='ignore')
tables = re.findall(r'(?m)^Table\s+([A-Za-z0-9_]+)\s*\{', code)
tpg_tables = sorted([t for t in tables if t.lower().startswith('tpg_')])
print('Tables tpg_*:', len(tpg_tables))
print(tpg_tables[:10])
```
